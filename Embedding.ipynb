```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
corpus = ["machine learning is great","machine learning is powerful"]
unigram_vectorizer = CountVectorizer(ngram_range=(1, 1))
unigram_X = unigram_vectorizer.fit_transform(corpus)
df_unigram = pd.DataFrame(unigram_X.toarray(), columns=unigram_vectorizer.get_feature_names_out())
print(df_unigram.columns.tolist())
```

```python
bigram_vectorizer = CountVectorizer(ngram_range=(2, 2))
bigram_X = bigram_vectorizer.fit_transform(corpus)
df_bigram = pd.DataFrame(bigram_X.toarray(), columns=bigram_vectorizer.get_feature_names_out())
print(df_bigram.columns.tolist())
```

```python
trigram_vectorizer = CountVectorizer(ngram_range=(3, 3))
trigram_X = trigram_vectorizer.fit_transform(corpus)
df_trigram = pd.DataFrame(trigram_X.toarray(), columns=trigram_vectorizer.get_feature_names_out())
print(df_trigram.columns.tolist())
```

```python
combined_vectorizer = CountVectorizer(ngram_range=(2, 3))
combined_X = combined_vectorizer.fit_transform(corpus)
df_combined = pd.DataFrame(
    combined_X.toarray(),
    columns=combined_vectorizer.get_feature_names_out()
)
print(df_combined.columns.tolist())
```

```python
from sklearn.feature_extraction.text import CountVectorizer
sentences = ["The cat sat on the mat.","The cat sat near the mat.","The cat played with a ball."]
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(sentences)
print("Vocabulary:")
print(vectorizer.get_feature_names_out())
print("\nBag of Words Matrix:")
print(X.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
documents = ["The cat sat on the mat""Cats and dogs are great pets","The dog sat on the log"]
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=vectorizer.get_feature_names_out())
print(tfidf_df)
```

```python
from gensim.models import Word2Vec
sentencedata = [["the", "quick", "brown", "fox", "jumps", "over", "the", "lazy", "dog"],["yous,", "can", "train", "word2vec", "on", "any", "text", "data"]]
model = Word2Vec(sentences=sentencedata, vector_size=100,window=5, min_count=1, workers=4, sg=1)
model.save("word2vec.model")
loaded_model = Word2Vec.load("word2vec.model")
similar_words = loaded_model.wv.most_similar("fox")
print(similar_words)
```
